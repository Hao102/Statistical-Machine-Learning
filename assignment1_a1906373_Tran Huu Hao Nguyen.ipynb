```python
# do not change the signature for this function
def linreg_compute_cost(X, Y, W): 
    """
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      Y (ndarray (m,1)) : target values
      W (ndarray (n,1)) : model parameters  
      
    Returns:
      J (scalar): cost
    """
# ====================== YOUR CODE HERE ======================  

## Firstly, we define the formula of the cost function J = 1/2m * (h(x) -y)^2 from seminar slide lecture week 2, page 9.

## Next, we will consider the number of training examples m
    
    m = X.shape[0] 

## Next, we will proceed to calculate with the predicted output: h(X) = X * W

    predicted_output_require = X @ W  

## Next, we will calculate the error vector by subtracting the target values ​​from the predicted output.
    error_finding = predicted_output_require - Y  

## Finally, we calculate the cost function J by the mean squared error formula with 1/(2*m)

    J = (1 / (2 * m)) * np.sum( error_finding ** 2)  

    
# ============================================================
    return J
```

```python
# do not change the signature for this function
def linreg_compute_gradient(X, Y, W): 
    """
    Computes the gradient for linear regression 
    Args:
      X : Data,m examples with n features
      Y : m target values
      W : n model parameters 
      
    Returns:
      dJ_dW : The gradient of the cost w.r.t. the parameters W, n values. 
    """
# ====================== YOUR CODE HERE ======================  

## From seminar slide lecture week 2, page 21, We will start from the cost function for vector linear regression: J(W) = (1/2m) * (XW - Y)^ T * (XW-Y) , to minimize the cost function, we calculate its gradient by taking the partial derivative of J for each 𝑊j, from which we have the formula (1/m) * X^T * (XW - Y), the formula is further derived from seminar slide week 2, page 21.

## Firstly, we will calculate the number of training examples m 

    m = X.shape[0]  

## Next, we will calculate the predicted output using the weights: h(X) = X * W   
    
    predicted_output_find = X @ W  

## Next, we will calculate the error vector by subtracting the target values ​​from the predicted output.
    
    error_vector =  predicted_output_find - Y 
    
## Finally, we calculate the gradient using the formula = (1/m) * (X^T * error ) 
    
    dJ_dW = (1 / m) * (X.T @ error_vector)
  

# ============================================================

    return dJ_dW
```

```python
# do not change the signature for this function
def linreg_gradient_descent(X, Y, W_in, cost_function, gradient_function, alpha, num_iters): 
    """
    Performs batch gradient descent to learn theta. Updates theta by taking 
    num_iters gradient steps with learning rate alpha
    
    Args:
      X                   : Data, n examples with m features
      Y                   : m target values
      W_in                : n initial model parameters  
      cost_function       : function to compute cost
      gradient_function   : function to compute the gradient
      alpha (float)       : Learning rate
      num_iters (int)     : number of iterations to run gradient descent
      
    Returns:
      W                   : final n values of parameters 
      J (scalar)          : final cost
      """
    
# ====================== YOUR CODE HERE ======================  

## Firstly, we have to perform the initial weight copy step so that we do not overwrite the initial input weight W_in and it will be updated step by step.     
    W = W_in.copy()  

## Next, we will iterate over a fixed number of gradient descent steps for i in the range (num_iters) and we can see that each step reduces the cost function a little bit
    for i in range(num_iters):
        ## Next we can calculate the gradient (derivative of the cost function with respect to the weights), we use again the formula of the gradient vector above which is (1/m) * X^T * (XW - Y) so we can calculate how much change is needed for each weight
        gradient_need = gradient_function(X, Y, W)
        
        ## Next, from seminar slide lecture week 2 page 17, we can apply the gradient descent formula W := W - α * ∇ J(W) to update the weights
        ## We will talk more about alpha which is the learning rate. If we see a small alpha, learning is slow (it may take more steps). And conversely, if we see a large alpha, learning is faster.
        W = W - alpha * gradient_need
    
    ## Finally, after all the iterations are done, we proceed to calculate the final cost function. From this thing, we can comment on how well the final weights fit the data.
    ## Final cost will be calculated by the formula J = (1/2m) sigma (XW - Y) ^2
    J = cost_function(X, Y, W)

    
# ===========================================================

    return W, J
```

```python
# initialize parameters
initial_W = np.ones(X_train.shape[1]).reshape(X_train.shape[1],1)

# gradient descent settings
iterations = 1000
alpha = 0.05

"""
Apply functions coded above to calculate final W and cost J
Use given datasets and parameters
"""
# ====================== YOUR CODE HERE ======================  

## We are given X_train and Y_train. And we will combine X_train (input features), Y_train (target values), initial_W (initial weights), linreg_compute_cost (cost function), linreg_compute_gradient (gradient function), alpha (learning rate) and iterations (number of updates) to find linreg_gradient_descent

W, J = linreg_gradient_descent(
    X_train,               
    Y_train,               
    initial_W,             
    linreg_compute_cost,   
    linreg_compute_gradient, 
    alpha,                 
    iterations             
)

# ============================================================


print("Please copy the folowing result into Question 2 in myUni Assignment Question 2")
print(np.round(np.sum(W)+J,2))
```

```python
# do not change the signature for this function
def sigmoid(z):
    """
    Compute sigmoid function given the input z.
    
    Parameters
    ----------
    z : array_like
    
    Returns
    -------
    g : array_like
        The computed sigmoid function. g has the same shape as z, since
        the sigmoid is computed element-wise on z.
        
    Instructions
    ------------
    Compute the sigmoid of each value of z (z can be a matrix, vector or scalar).
    """
    # convert input to a numpy array
    z = np.array(z).astype("float")
    
    # ====================== YOUR CODE HERE ======================  
    # DO NOT use any other import statements for this question

    ## From the week 4 seminar slide lecture page 10, we have the formula hw(z) = 1 / (1 + e^(-z)) and we apply it to find g:
    g = 1 / (1 + np.exp(-z))
   
    # =============================================================
    return g
```

```python
# do not change the signature for this function
def logreg_costFunctionReg(W, X, Y, lambda_):
    """
    Compute cost and gradient for logistic regression with L2 regularization.
    
    Parameters
    ----------
    W : Logistic regression vector of n parameters,
        where n is the number of features including intercept.
    
    X : The data set with shape (m,n). m is the number of examples, and
        n is the number of features.
    
    Y : The vector of data labels of size m.
    
    lambda_ : float
        The regularization parameter. 
    
    Returns
    -------
    J : float
        The computed value for the L2 regularized cost function. 
   
    """
    # Initialize some useful values
    m = Y.size  # number of training examples

    # You need to return the following variables correctly 
    J = 0

    # ===================== YOUR CODE HERE ======================

    ## Firstly, from seminar slide lecture week 4 page 10, in logistic regression function we will need to calculate z = W^T X but in python code we can convert it to z = XW to vectorize to process all samples at once
    ## Next, we prepare to force Y to column (m,1) to make matrix multiplication easier later, and we force column W to column (n,1) to be compatible when multiplying by X to calculate Z
    Y = np.array(Y).reshape(-1, 1)      # shape (m, 1)
    W = np.array(W).reshape(-1, 1)      # shape (n, 1)

    ## Then we calculate the dot product Z = X * W
    z = X @ W       ## shape (m,1)

    ## Next, we calculate the sigmoid function using the formula h(z) = 1 / ( 1 + e^-z)
    #h = 1 / (1 + np.exp(-z))
    h = sigmoid(z)

    ## Based on the formula given in the beginning of the problem, we will start calculating the cost function without regularization using the formula 1/m * (-Y log(h(X)) - (1-Y) log(1-h(X)))

    cost_function_no_regularization = (-Y * np.log(h)) - ((1 - Y) * np.log(1 - h))    ## shape(m,1)

    ## Next, we take the overall average over all sample m
    cost_need_no_regularization = (1 / m) * np.sum(cost_function_no_regularization)
   
    ## Next, we consider regularisation, based on the given formula, and regularisation only applies to W[1] so we exclude W[0] 

    cost_yes_regularisation = (lambda_ / (2 * m)) * np.sum(W[1:] ** 2)

    ## Finally, we sum up the final cost based on the formula above the question given.
    J = cost_need_no_regularization + cost_yes_regularisation 

    
    # =============================================================
    
    return J
```

```python
# do not change the signature for this function
def logreg_GradFunctionReg(W, X, Y, lambda_):
    """
    Compute cost and gradient for logistic regression with regularization.
    
    Parameters
    ----------
    W : Logistic regression vector of n parameters,
        where n is the number of features including intercept.
    
    X : The data set with shape (m,n). m is the number of examples, and
        n is the number of features.
    
    Y : The vector of data labels of size m.
    
    lambda_ : float
        The regularization parameter. 
    
    Returns
    -------
    grad : A vector of size n which is the gradient of the cost
        function with respect to W, at the current values of theta.
    
    """
    # Initialize some useful values
    m = Y.size  # number of training examples

    # You need to return the following variables correctly 
    grad = np.zeros(W.shape)

    # ===================== YOUR CODE HERE ======================

    ## Firstly, we need to make sure that X, W are column vectors to perform matrix multiplication correctly.
    Y = np.array(Y).reshape(-1, 1)      # shape (m, 1)
    W = np.array(W).reshape(-1, 1)      # shape (n, 1)

    ## Combine knowledge about gradient descent for Logistic Regression seminar slides lecture week 4 pages 15 and 16, and the given J(W) function
    ## We can see the derivative gradient function mathematically as follows: ∇J(W) = (1/m) * X^T(h - Y) + (λ/m) * [0, w1, w2, ..., wn] with h(z) = 1 / ( 1 + e^-z)

    ## Next, we calculate z = w^T X , but in python code we can write it as z = XW, it is also called logit score
    z = X @ W  # shape (m, 1)

    ## Next, we calculate h^{(z)} = 1 / (1 + e^{-z}), that is also the sigmoid formula.
    #h = 1 / (1 + np.exp(-z))         # shape(m, 1)
    h = sigmoid(z)
    ## Next, we will consider the prediction error, which is also ∂J/∂z: the derivative of the sigmoid part of the loss.

    error_predict = h - Y        # shape(m, 1)

    ## Next, based on the semiar slide lecture week 4 page 16, we calculate the Gradient from the loss part on the formula ∇J_loss = (1/m) ∑ (h - y) * x_j or on the calculation code is (1/m) * X.T * error_predict
    find_gradient_loss = (X.T @ error_predict) / m       # shape (n, 1)

    ## Next, we calculate the gradient from the regularization part, the formula is (λ/m) * W, but W[0] ( W[0] - the intercept) is not regularized so it = 0

    regularization_part = (lambda_ / m) * W
    regularization_part[0] = 0 

    ## Finally, we calculate the total gradient we need by adding the gradients from los and from regularization

    grad = find_gradient_loss + regularization_part       # shape (n, 1)

    ## Convert grad to 1D
    grad = np.ravel(grad)
    
    # =============================================================
    
    return grad
```

```python
# do not change the signature for this function
def logreg_gradient_descent_reg(X, Y, W, cost_function, gradient_function, alpha, num_iters, lambda_): 
    """
    Performs batch gradient descent to learn theta. Updates theta by taking 
    num_iters gradient steps with learning rate alpha
    
    Args:
      X                   : Data ndarray array, n examples with m features
      Y                   : ndarray vector of target n values
      W                   : ndarray vector of initial m model parameters 
      cost_function       : function to compute cost
      gradient_function   : function to compute the gradient
      alpha (float)       : Learning rate
      num_iters (int)     : number of iterations to run gradient descent
      
    Returns:
      W (ndarray)         : Updated values of parameters 
      """
    
    # ===================== YOUR CODE HERE ======================
    W = np.array(W).reshape(-1, 1)
    ## We apply the formula from gradient descent for log reg (Regularized) from seminar slide lecture week 4 page 28, W:=W - alpha ∇ J(W)
    ## W:=W - alpha ∇ J(W) where alpha is the learning rate and ∇ J(W) is the gradient of the cost function
    ## Firstly, we proceed to use looping over for the number of iterations:
    for i in range(num_iters):
    ## Next, we start computing the gradient ∇J(W). In this problem, this function consists of two parts, the gradient of the logistic loss and the gradient of the regularization term.
        gradient_needed_updating = gradient_function(W, X, Y, lambda_) 
    ## Next, our goal is to minimize the cost so we have to move against the gradient, so we update the weights again, W := W - α * ∇J(W)
        W = W - alpha * gradient_needed_updating.reshape(-1, 1)

    ## Finally, after optimizing the weights, we continue to calculate the final cost;
    J = cost_function(W, X, Y, lambda_)
    
    # =============================================================
    
    return J, W
```

```python
import numpy as np

## Firstly, we set up the dataset and weights according to the given question data.

## # X is a 3x5 matrix
X = np.array([
    [1, 1, 1, 1, 1],     
    [3, 6, 7, 8, 11],    
    [4, 7, 8, 9, 37]    
])

# Weight vector
w = np.array([
    [13],
    [8],
    [6]
])

## To begin with, we will continue with Eq1 based on the given data given that sum i <Xi ^T W> when X belongs to R(d,1) and W belongs to R(d,1)

## We create a variable to record the result of Eq1

eq1_result_needed = 0

## Next, we loop through each column of X
for i in range(X.shape[1]):
    ## Next, we reshape x_i_using into a (3x1) column vector
    x_i_using = X[:, i].reshape(-1, 1)
    ## Next, from seminar slide lecture week 1 page 44, we calculate the matrix product x_i^T * W and we will have x_i_using ^T (1,3) * W(3,1) = (1,1)
    x_i_dot_product = x_i_using.T @ w
    ## Then we will extract the scalar number from the (1,1) matrix we just calculated above.
    eq1_result_needed = eq1_result_needed +  x_i_dot_product[0, 0]

print("Eq1 result:", eq1_result_needed)


## Secondly, we will come to expression 1 with the form sum_i <W, X_i>

## We create a variable to record the result of expression 1
case1_result = 0
## Next, we loop through each column of X
for i in range(X.shape[1]):
## Next, we will take x_i_using as a 1-dimensional vector
    x_i_using = X[:, i]
## Next, from seminar slide lecture week 1 page 44, we will use the dot product between W and x_i_using
    case1_dot_product =  np.dot(w.flatten(), x_i_using)
## Then, we will accumulate the result
    case1_result = case1_result + case1_dot_product

print("Expression 1:", case1_result)


## Thirdly, we will come to expression 2 with the form (sum_i X_i) * W
## Since this is a special case, we will do something different to clarify this case.
try:
## We will calculate the sum on the columns of X and we get the shape X(3, 1)
    sum_x_finding = np.sum(X, axis=1).reshape(-1, 1)
## Next, from seminar slide lecture week 1 page 44, we will continue to multiply the matrix between the sum on columns X and W, 
## The result obtained for that shape is sum_x_finding (3, 1) * W (3, 1) -> invalid
    case2_result = sum_x_finding @ w
    print("Expression 2", case2_result)
except ValueError as e:
    print("Expression 2 is not equivalent to Eq1:", e)


## Fouthly, we will come to expression 2 with the form sum_i Tr(X_i W^T)
## This case is also a special case when using the trace (sum of diagonal elements) of the matrix X_i W^T
## We create a variable to record the result of expression 3
case3_result = 0
## Next, we loop through each column of X
for i in range(X.shape[1]):
## Next, we will reshape x_i_using to (3,1) in this case 3
    x_i_using = X[:, i].reshape(-1, 1)   
## Next, from seminar slide lecture week 1 page 46, we will calculate the outer product by taking x_i_using *( w^T) and considering the shape we will see that x_i_using (3, 1) * w^T (1, 3) = (3, 3)
    out_product = x_i_using @ w.T  
## Next, we will proceed to calculate the diagonal product
    trace_total_needed = np.trace(out_product)    
## Then, we will accumulate the result
    case3_result = case3_result + trace_total_needed

print("Expression 3:", case3_result)



## Fifthly, we will come to expression 2 with the form W^T * (sum_i X_i)
## From seminar slide lecture week 1 page 46 and 44, For this expression, we sum all the vectors X_i, then multiply by W^T
sum_x_finding = np.sum(X, axis=1).reshape(-1, 1)    # shape (3, 1)
## Next, we do the multiplication between sum_x_finding and w^T, where in terms of shape we see that w^T (1, 3) * sum_x_finding (3, 1) = (1, 1)
product_dot_finding = w.T @ sum_x_finding
## Then, we will accumulate the result
case4_result = product_dot_finding[0, 0]

print("Expression 4:", case4_result)

```
